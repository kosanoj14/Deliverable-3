```python
# Display Dice

d6 = Dice(6)
d20 = Dice(20)
print(f"Terning-representasjon: {d6}")
print(f"Terningkast med roll(): {d6.roll()}")
print(f"Terningkast med __call__: {d20()}")

# Opprett og test Character
char = Character("Bonde", hp_max=50, damage=8)
print(f"\n{char}")
char.take_damage(20)
print(f"HP etter skade: {char.hp}, lever fremdeles: {char.is_alive}")
char.heal(10)
print(f"HP etter healing: {char.hp}")
```

```python
print("Base methods and attributes for characters")

base = Character("basi")

print("Character")
print("Attributes:")
print(base.__dict__)

print("\nMethods:")
for method in dir(base):
    if not method.startswith("__"):
        print(method)

```

```python
print("Methods and attributes for a Mage")
obj = Mage("magic")

print("Mage")
print("Attributes:")
print(obj.__dict__)

print("\nMethods:")
for method in dir(obj):
    if not method.startswith("__"):
        print(method)
```

```python
print("Methods and attributes for a Warrior")
job = Warrior("krig")
print("Warrior")
print("Attributes:")
print(job.__dict__)

print("\nMethods:")
for method in dir(job):
    if not method.startswith("__"):
        print(method)
```

```python
# Battle simulations

std = Character("name")
std_with1 = Character("Peasant 1", weapon=sword)
std_with2 = Character("Peasant 2", weapon=sword)
war_no = Warrior("war1")
mag_no = Mage("mag1")
war_yes = Warrior("war2", weapon=sword)
mag_yes = Mage("mag2", weapon=carrot_arrow)

print("-- battle with no weapons (warrior vs character) --")
battle1 = Battle(std, war_no)
battle1.simulate_battle()

print("-- battle with no weapons (warrior vs mage) --")
battle2 = Battle(war_no, mag_no)
battle2.simulate_battle()

print("-- battle with weapons (mage vs warrior) --")
battle3 = Battle(mag_yes, war_yes)
battle3.simulate_battle()

print("-- battle with weapons (character vs character) --")
battle4 = Battle(std_with1, std_with2)
battle4.simulate_battle()
print("Here we se the dice roll change the result of the battle more often")
```
